```python
import matplotlib.pyplot as plt
import pandas as pd
import os
```

```python
folder_names = [
    "/Users/arnavmehta/homework_spring2026/hw2/exp/CartPole-v0_cartpole_sd1_20260222_124907/log.csv",
    "/Users/arnavmehta/homework_spring2026/hw2/exp/CartPole-v0_cartpole_rtg_sd1_20260222_130222/log.csv",
    "/Users/arnavmehta/homework_spring2026/hw2/exp/CartPole-v0_cartpole_na_sd1_20260222_130253/log.csv",
    "/Users/arnavmehta/homework_spring2026/hw2/exp/CartPole-v0_cartpole_rtg_na_sd1_20260222_130334/log.csv"
]

labels = [
    "Vanilla PG",
    "Reward-to-Go",
    "No Advantage Norm",
    "Reward-to-Go + Adv Norm"
]

plt.figure(figsize=(15,6))

for path, label in zip(folder_names, labels):
    df = pd.read_csv(path)
    x = df["Train_EnvstepsSoFar"]
    y = df["Eval_AverageReturn"]
    plt.plot(x, y, label=label)

plt.xlabel("Environment Steps")
plt.ylabel("Evaluation Average Return")
plt.title("CartPole-v0: Policy Gradient Variants (Small Batch)")

# Legend outside on the right
plt.legend(loc="center left", bbox_to_anchor=(1, 0.5))

plt.tight_layout()
plt.show()
    
```
